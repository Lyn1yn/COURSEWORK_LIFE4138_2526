### Question 3

Use three different methods to print the numbers 200 to 300.

````python
# method 1
for i in range (200, 301):
    print(i)

# method 2
num = 200
while num <= 300:
    print(num)
    num += 1

# method 3
def print_number(n):
    if n > 300:
        return
    print(n)
    print_number(n+1)
print_number(200)



````

````python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

diamonds = sns.load_dataset('diamonds')
diamonds.head()

#1
diamonds.plot.scatter(x='carat', y='price')
plt.title('The relationship between carat and price')
plt.xlabel('Carat')
plt.ylabel('Price')
plt.show()

#2
diamonds['price'].plot.hist()
plt.title('Distribution of prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()

#3
diamonds.boxplot(column='price', by='cut', grid=False)
plt.title('Price distribution by cut')
plt.suptitle('')
plt.xlabel('cut')
plt.ylabel('price')
plt.show()
````

### Question 10
Write a function that will convert carat to grams and find the heaviest diamond.

1 carat = 0.2 grams

````python
import seaborn as sns
import pandas as pd

diamonds = sns.load_dataset('diamonds')
def convert(carat):
    return carat*0.2
diamonds['grams'] = diamonds['carat'].apply(convert)
diamonds.head()

#heaviest:1.002gram
diamonds.sort_values(by = 'grams', ascending = False)
````
